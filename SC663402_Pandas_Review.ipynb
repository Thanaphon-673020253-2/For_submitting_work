### โจทย์ 2.1
สร้าง Series ชื่อ `rain_by_city` ซึ่งเก็บปริมาณฝนรวมของแต่ละเมือง โดยกำหนดให้ index เป็นชื่อเมือง จากนั้นตอบคำถามต่อไปนี้ด้วยโปรแกรม

1. เมืองที่มีปริมาณฝนรวมสูงสุด 3 อันดับแรก
2. สัดส่วนปริมาณฝนของแต่ละเมืองเทียบกับปริมาณฝนรวมทั้งหมด แสดงเป็นร้อยละทศนิยม 2 ตำแหน่ง
3. จำนวนเมืองที่มีปริมาณฝนรวมสูงกว่าค่าเฉลี่ยของทุกเมือง

````python
# คำตอบข้อ 2.1
# สร้าง Series ชื่อ rain_by_city เก็บปริมาณฝนรวมของแต่ละเมือง โดย index เป็นชื่อเมือง
rain_by_city = df.groupby("city")["rainfall_mm"].sum()

# 1. เมืองที่มีปริมาณฝนรวมสูงสุด 3 อันดับแรก
top3_rain = rain_by_city.nlargest(3)
print("1. เมืองที่มีปริมาณฝนรวมสูงสุด 3 อันดับแรก:")
print(top3_rain)

# 2. สัดส่วนปริมาณฝนของแต่ละเมืองเทียบกับปริมาณฝนรวมทั้งหมด แสดงเป็นร้อยละทศนิยม 2 ตำแหน่ง
rain_share_pct = ((rain_by_city / rain_by_city.sum()) * 100).round(2)
print("\n2. สัดส่วนปริมาณฝนของแต่ละเมืองเทียบกับปริมาณฝนรวมทั้งหมด (%):")
print(rain_share_pct)

# 3. จำนวนเมืองที่มีปริมาณฝนรวมสูงกว่าค่าเฉลี่ยของทุกเมือง
avg_rain = rain_by_city.mean()
n_above_avg = (rain_by_city > avg_rain).sum()
print(f"\n3. ค่าเฉลี่ยปริมาณฝนรวมของทุกเมือง = {avg_rain:.2f} มม.")
print(f"จำนวนเมืองที่มีปริมาณฝนรวมสูงกว่าค่าเฉลี่ยของทุกเมือง: {n_above_avg} เมือง")
````

`city`, `country`, `region`, `n_records`, `avg_temp`, `total_rain`

### โจทย์ 3.2
เปรียบเทียบปริมาณหน่วยความจำที่ใช้ระหว่างตารางที่นำเข้าตามปกติกับตารางจากโจทย์ 3.1 ด้วย `.memory_usage(deep=True)`
รายงานผลเป็นร้อยละที่ลดลง และ **อธิบาย** ว่าคอลัมน์ใดมีผลต่อการลดลงมากที่สุด เพราะเหตุใด

````python
# คำตอบข้อ 3.2
# เปรียบเทียบปริมาณหน่วยความจำด้วย .memory_usage(deep=True)
mem_orig = df.memory_usage(deep=True).sum()
mem_opt = df_custom.memory_usage(deep=True).sum()
reduction_pct = ((mem_orig - mem_opt) / mem_orig) * 100

print(f"ขนาดหน่วยความจำตารางที่นำเข้าปกติ: {mem_orig:,} bytes")
print(f"ขนาดหน่วยความจำตารางจากโจทย์ 3.1: {mem_opt:,} bytes")
print(f"ปริมาณหน่วยความจำที่ลดลง: {reduction_pct:.2f}%")

print("\n[อธิบาย]:")
print("1. คอลัมน์ที่มีผลต่อการลดลงของหน่วยความจำมากที่สุด คือ คอลัมน์ที่ไม่ได้เลือกอ่าน (usecols) และคอลัมน์ที่เป็นข้อความซ้ำๆ เช่น 'city' และ 'region'")
print("2. สาเหตุที่หน่วยความจำลดลงมาก เกิดจาก 2 ปัจจัยหลัก:")
print("   - การเลือกอ่านเฉพาะคอลัมน์ที่จำเป็น (usecols) ทำให้ไม่ต้องจองพื้นที่หน่วยความจำสำหรับคอลัมน์อื่นที่ไม่เกี่ยวข้อง")
print("   - การกำหนดชนิดข้อมูลเป็น 'category' สำหรับ 'city' และ 'region' ช่วยลดหน่วยความจำได้อย่างมหาศาล เพราะ category จะเก็บข้อความจริงเพียงชุดเดียว แล้วใช้รหัสตัวเลขจำนวนเต็ม (integer codes) ในการอ้างอิงแทนการเก็บสตริงซ้ำๆ ในทุกแถว ซึ่งสตริงใน Python มี overhead ของ metadata สูงมาก")
````

### โจทย์ 3.3
ใช้การอ่านข้อมูลแบบ `chunksize` เพื่อคำนวณค่าต่อไปนี้ โดยห้ามโหลดข้อมูลทั้งไฟล์เข้าหน่วยความจำพร้อมกัน และห้ามใช้ `pd.concat`

1. อุณหภูมิสูงสุดและต่ำสุดของทั้งไฟล์
2. อุณหภูมิเฉลี่ยรายเมือง

**อธิบาย:** เหตุใดการนำค่าเฉลี่ยของแต่ละก้อนมาเฉลี่ยกันอีกครั้งจึงให้ผลไม่ถูกต้อง และควรสะสมค่าใดแทนจึงจะได้ผลที่ถูกต้อง

````python
# คำตอบข้อ 3.3
# คำนวณด้วย chunksize โดยไม่โหลดข้อมูลทั้งไฟล์พร้อมกันและไม่ใช้ pd.concat
global_min = float("inf")
global_max = float("-inf")
city_temp_sum = {}
city_temp_count = {}

for chunk in pd.read_csv(BASE + "world_weather_sample.csv", chunksize=500):
    valid_temps = chunk["temperature_c"].dropna()
    if not valid_temps.empty:
        global_min = min(global_min, valid_temps.min())
        global_max = max(global_max, valid_temps.max())
    
    # รวม sum และ count ของแต่ละเมืองในแต่ละ chunk
    grp = chunk.groupby("city")["temperature_c"].agg(["sum", "count"])
    for city, row in grp.iterrows():
        city_temp_sum[city] = city_temp_sum.get(city, 0.0) + row["sum"]
        city_temp_count[city] = city_temp_count.get(city, 0) + row["count"]

city_avg_temp = pd.Series({c: city_temp_sum[c] / city_temp_count[c] for c in city_temp_sum})

print("1. อุณหภูมิของทั้งไฟล์:")
print(f"   อุณหภูมิต่ำสุด = {global_min} °C, อุณหภูมิสูงสุด = {global_max} °C")
print("\n2. อุณหภูมิเฉลี่ยรายเมือง (5 เมืองแรก):")
print(city_avg_temp.head())

print("\n[อธิบาย]:")
print("เหตุใดการนำค่าเฉลี่ยของแต่ละก้อนมาเฉลี่ยกันอีกครั้งจึงไม่ถูกต้อง:")
print("เพราะในแต่ละ chunk แต่ละเมืองอาจมีจำนวนแถวที่มีข้อมูลไม่เท่ากัน (Sample size / Weight ต่างกัน) การนำค่าเฉลี่ยย่อยมาเฉลี่ยกันตรงๆ ถือเป็น Unweighted Average ซึ่งจะให้น้ำหนักแต่ละ chunk เท่ากัน ทำให้ผลลัพธ์คลาดเคลื่อน")
print("สิ่งที่ควรสะสมแทน: ต้องสะสม 'ผลรวมค่าอุณหภูมิ (sum)' และ 'จำนวนข้อมูลจริงที่ไม่ใช่ค่าว่าง (count)' ของแต่ละเมืองแยกไว้ แล้วนำ sum รวมทั้งหมดมาหารด้วย count รวมทั้งหมดในตอนท้าย จึงจะได้ค่าเฉลี่ยรวมที่ถูกต้องตามหลักสถิติ")
````

1. จำนวนแถวที่ `temperature_c` อยู่นอกช่วง -60 ถึง 60

````python
# คำตอบข้อ 4.2
# 1. จำนวนแถวที่ temperature_c อยู่นอกช่วง -60 ถึง 60
cond_temp = (~df["temperature_c"].between(-60, 60)) & df["temperature_c"].notna()
n_temp_out = cond_temp.sum()

# 2. จำนวนแถวที่ humidity_pct อยู่นอกช่วง 0 ถึง 100
cond_humid = (~df["humidity_pct"].between(0, 100)) & df["humidity_pct"].notna()
n_humid_out = cond_humid.sum()

# 3. จำนวนแถวที่มีความผิดปกติอย่างน้อยหนึ่งคอลัมน์ (ระวังการนับซ้ำด้วยตัวดำเนินการ |)
cond_either = cond_temp | cond_humid
n_either_out = cond_either.sum()

print(f"1. จำนวนแถวที่ temperature_c ผิดปกติ (นอกช่วง -60 ถึง 60): {n_temp_out} แถว")
print(f"2. จำนวนแถวที่ humidity_pct ผิดปกติ (นอกช่วง 0 ถึง 100): {n_humid_out} แถว")
print(f"3. จำนวนแถวที่มีความผิดปกติอย่างน้อยหนึ่งคอลัมน์ (ไม่นับซ้ำ): {n_either_out} แถว")
````

### โจทย์ 6.1
เขียนเงื่อนไขสำหรับกรองข้อมูลต่อไปนี้

1. ข้อมูลในภูมิภาคเอเชียที่ปริมาณฝนมากกว่า 5 มิลลิเมตร และความเร็วลมมากกว่า 20 กิโลเมตรต่อชั่วโมง
2. ข้อมูลที่ไม่ใช่เมือง Bangkok, Tokyo และ London และมีความชื้นระหว่าง 40 ถึง 60
3. แถวที่มีค่าว่างตั้งแต่ 2 คอลัมน์ขึ้นไป
4. แถวที่ชื่อเมืองมีคำว่า New หรือ San ประกอบอยู่ โดยใช้ `.str.contains` ร่วมกับ regular expression

````python
# คำตอบข้อ 6.1
# 1. ภูมิภาคเอเชีย ฝน > 5 mm และ ลม > 20 km/h
q1 = df[(df["region"] == "Asia") & (df["rainfall_mm"] > 5) & (df["wind_speed_kmh"] > 20)]
print(f"1. เอเชีย ฝน>5 ลม>20: พบ {len(q1)} แถว")

# 2. ไม่ใช่ Bangkok, Tokyo, London และ ความชื้น 40-60
q2 = df[(~df["city"].isin(["Bangkok", "Tokyo", "London"])) & (df["humidity_pct"].between(40, 60))]
print(f"2. ไม่ใช่ 3 เมืองดังกล่าว และความชื้น 40-60: พบ {len(q2)} แถว")

# 3. แถวที่มีค่าว่างตั้งแต่ 2 คอลัมน์ขึ้นไป
q3 = df[df.isna().sum(axis=1) >= 2]
print(f"3. แถวที่มีค่าว่างตั้งแต่ 2 คอลัมน์ขึ้นไป: พบ {len(q3)} แถว")

# 4. เมืองที่มีคำว่า New หรือ San (regex)
q4 = df[df["city"].str.contains(r"New|San", regex=True)]
print(f"4. เมืองที่มี New หรือ San: พบ {len(q4)} แถว (เมือง: {q4['city'].unique().tolist()})")
````

### โจทย์ 6.2
ใช้ `.where()` และ `.mask()` สร้างคอลัมน์ใหม่สองคอลัมน์ ได้แก่ คอลัมน์อุณหภูมิที่แทนค่าผิดปกติด้วยค่าว่าง และคอลัมน์ปริมาณฝนที่จำกัดค่าสูงสุดไว้ที่ค่าเปอร์เซ็นไทล์ที่ 99
**อธิบาย:** แนวทางนี้แตกต่างจากการลบแถวทิ้งอย่างไรในแง่ของการวิเคราะห์ในขั้นถัดไป

````python
# คำตอบข้อ 6.2
# สร้างคอลัมน์อุณหภูมิที่แทนค่าผิดปกติ (นอกช่วง -60 ถึง 60) ด้วยค่าว่าง ด้วย .where()
df["temp_cleaned"] = df["temperature_c"].where(df["temperature_c"].between(-60, 60))

# สร้างคอลัมน์ปริมาณฝนที่จำกัดค่าสูงสุดไว้ที่เปอร์เซ็นไทล์ที่ 99 ด้วย .mask()
p99_rain = df["rainfall_mm"].quantile(0.99)
df["rain_capped"] = df["rainfall_mm"].mask(df["rainfall_mm"] > p99_rain, p99_rain)

print(f"เปอร์เซ็นไทล์ที่ 99 ของปริมาณฝน = {p99_rain:.2f} มม.")
print(f"จำนวนค่าว่างใน temp_cleaned = {df['temp_cleaned'].isna().sum()} (เดิมมี {df['temperature_c'].isna().sum()})")
print(f"ค่าสูงสุดของ rain_capped หลังจำกัดค่า = {df['rain_capped'].max():.2f} มม. (เดิมคือ {df['rainfall_mm'].max():.2f} มม.)")

print("\n[อธิบาย]:")
print("แนวทางนี้แตกต่างจากการลบแถวทิ้ง (dropna) อย่างไร:")
print("1. การลบแถวทิ้งจะทำให้สูญเสียข้อมูลทั้งหมดของแถวนั้นไปทันที (เช่น ความเร็วลม ความกดอากาศ วันที่) ซึ่งทำให้จำนวน observation ลดลงและทำลายความต่อเนื่องของ Time Series")
print("2. การใช้ .where() แทนค่าผิดปกติด้วย NaN จะช่วยกำจัดผลกระทบของค่า Outlier ในการวิเคราะห์อุณหภูมิ แต่ยังคงรักษาข้อมูลในคอลัมน์อื่นที่มีประโยชน์ไว้ได้ครบถ้วน")
print("3. การใช้ .mask() จำกัดค่าสูงสุด (Winsorization) ช่วยลดผลกระทบของค่าสุดโต่งไม่ให้ดึงค่าสถิติ (mean/variance) สูงเกินจริง โดยไม่จำเป็นต้องตัดแถวทิ้ง")
````

### โจทย์ 6.3
หาวันที่มีอุณหภูมิห่างจากค่าเฉลี่ยของเมืองตนเองมากที่สุดของแต่ละเมือง
ผลลัพธ์ต้องมีหนึ่งแถวต่อหนึ่งเมือง ประกอบด้วยคอลัมน์ `city`, `date`, `temperature_c`, `deviation`

````python
# คำตอบข้อ 6.3
# คำนวณค่าเฉลี่ยของเมืองตนเอง และหาค่าความต่างสัมบูรณ์ (deviation)
city_avg = df.groupby("city")["temperature_c"].transform("mean")
df["deviation"] = (df["temperature_c"] - city_avg).abs()

# เลือกแถวที่มี deviation สูงสุดในแต่ละเมือง
idx_max = df.groupby("city")["deviation"].idxmax()
res_63 = df.loc[idx_max, ["city", "date", "temperature_c", "deviation"]].sort_values("deviation", ascending=False).reset_index(drop=True)

print("วันที่มีอุณหภูมิห่างจากค่าเฉลี่ยของเมืองตนเองมากที่สุดของแต่ละเมือง (23 เมือง):")
print(res_63)
````

### โจทย์ 7.2
ตอบคำถามต่อไปนี้ด้วยโปรแกรม

1. ข้อมูลครอบคลุมทั้งหมดกี่วัน และมีวันใดในช่วงดังกล่าวที่ไม่มีข้อมูลเลยหรือไม่ โดยใช้ `pd.date_range` ประกอบการตรวจสอบ
2. อุณหภูมิเฉลี่ยของวันหยุดสุดสัปดาห์เปรียบเทียบกับวันธรรมดา จำแนกตามภูมิภาค
3. เดือนที่มีปริมาณฝนรวมสูงสุดของแต่ละเมือง

````python
# คำตอบข้อ 7.2
# 1. ข้อมูลครอบคลุมกี่วัน และมีวันใดไม่มีข้อมูลเลยหรือไม่
all_days = pd.date_range(start=df["date"].min(), end=df["date"].max())
unique_days = df["date"].unique()
missing_days = all_days.difference(unique_days)

print(f"1. ข้อมูลครอบคลุมทั้งหมด {len(all_days)} วัน (ตั้งแต่ {df['date'].min().date()} ถึง {df['date'].max().date()})")
print(f"   จำนวนวันที่ไม่มีข้อมูลเลยในช่วงดังกล่าว: {len(missing_days)} วัน")

# 2. อุณหภูมิเฉลี่ยวันหยุดสุดสัปดาห์ vs วันธรรมดา จำแนกตามภูมิภาค
weekend_summary = df.groupby(["region", "is_weekend"])["temperature_c"].mean().unstack()
weekend_summary.columns = ["วันธรรมดา (Weekday)", "วันหยุด (Weekend)"]
print("\n2. อุณหภูมิเฉลี่ยวันหยุด vs วันธรรมดา จำแนกตามภูมิภาค:")
print(weekend_summary.round(2))

# 3. เดือนที่มีปริมาณฝนรวมสูงสุดของแต่ละเมือง
rain_by_month = df.groupby(["city", "year_month"])["rainfall_mm"].sum().reset_index()
max_rain_month = rain_by_month.loc[rain_by_month.groupby("city")["rainfall_mm"].idxmax()].reset_index(drop=True)
print("\n3. เดือนที่มีปริมาณฝนรวมสูงสุดของแต่ละเมือง (5 เมืองแรก):")
print(max_rain_month.head())
````

### โจทย์ 7.3
ใช้ `.resample()` สรุปข้อมูลของเมือง Bangkok เป็นราย 7 วัน โดยแสดงอุณหภูมิเฉลี่ยและปริมาณฝนรวมในตารางเดียวกัน
**อธิบาย:** `.resample()` แตกต่างจาก `groupby(df["date"].dt.month)` อย่างไร และกรณีใดที่จำเป็นต้องใช้ `.resample()` เท่านั้น

````python
# คำตอบข้อ 7.3
bkk_7d = (
    df[df["city"] == "Bangkok"]
    .set_index("date")
    .sort_index()
    .resample("7D")
    .agg(
        avg_temp=("temperature_c", "mean"),
        total_rain=("rainfall_mm", "sum"),
    )
)

print("สรุปข้อมูลเมือง Bangkok ราย 7 วัน ด้วย .resample():")
print(bkk_7d.round(2))

print("\n[อธิบาย]:")
print("1. ความแตกต่างระหว่าง .resample() กับ groupby(dt.month):")
print("   - groupby(dt.month) สรุปข้อมูลตามคุณลักษณะปฏิทิน (เดือน 1, 2, 3) โดยรวบวันทั้งหมดในเดือนเดียวกันเข้าด้วยกันโดยไม่สนใจความต่อเนื่องของแกนเวลา")
print("   - .resample() สร้างช่วงเวลาที่มีความถี่คงที่ตามแกนเวลาจริง (เช่น ราย 7 วัน) และหากช่วงเวลาใดไม่มีข้อมูล จะสร้างแถวไว้ (เป็น NaN) ทำให้รักษาความต่อเนื่องของไทม์ไลน์ได้สมบูรณ์")
print("2. กรณีที่จำเป็นต้องใช้ .resample() เท่านั้น:")
print("   - เมื่อต้องการจัดกลุ่มตามความถี่เวลาที่ไม่ตรงกับฟิลด์ปฏิทินมาตรฐาน เช่น ทุกๆ 7 วัน ('7D'), ทุก 2 สัปดาห์ หรือรายชั่วโมง")
print("   - เมื่อต้องการทำ Downsampling/Upsampling ในการวิเคราะห์อนุกรมเวลา (Time Series Forecasting / Resampling)")
````

### โจทย์ 8.1
ตอบคำถามต่อไปนี้ด้วยโปรแกรม

1. คอลัมน์ใดมีค่าว่างมากที่สุด และคิดเป็นร้อยละเท่าใด
2. ค่าว่างกระจายตัวเท่ากันทุกเมืองหรือไม่ ให้สร้างตารางแสดงสัดส่วนค่าว่างของ `temperature_c` จำแนกตามเมือง
3. หากลบทุกแถวที่มีค่าว่างด้วย `dropna()` จะเหลือข้อมูลคิดเป็นร้อยละเท่าใดของข้อมูลเดิม และเมืองใดสูญเสียข้อมูลมากที่สุด

````python
# คำตอบข้อ 8.1
# 1. คอลัมน์ที่มีค่าว่างมากที่สุด และคิดเป็นร้อยละเท่าใด
missing_counts = df.isna().sum()
max_na_col = missing_counts.idxmax()
max_na_pct = (df[max_na_col].isna().mean() * 100).round(2)
print(f"1. คอลัมน์ที่มีค่าว่างมากที่สุดคือ '{max_na_col}' มีค่าว่าง {missing_counts[max_na_col]} แถว (คิดเป็น {max_na_pct}%)")

# 2. สัดส่วนค่าว่างของ temperature_c จำแนกตามเมือง
temp_na_by_city = (df.groupby("city")["temperature_c"].apply(lambda s: s.isna().mean()) * 100).round(2)
print("\n2. สัดส่วนค่าว่างของ temperature_c จำแนกตามเมือง (%):")
print(temp_na_by_city)
print(f"สรุป: ค่าว่างไม่ได้กระจายตัวเท่ากันทุกเมือง (มีตั้งแต่ {temp_na_by_city.min()}% ถึง {temp_na_by_city.max()}%)")

# 3. หากลบทุกแถวที่มีค่าว่างด้วย dropna()
df_drop = df.dropna()
remain_pct = (len(df_drop) / len(df)) * 100
rows_lost = df.groupby("city").size() - df_drop.groupby("city").size()
worst_city = rows_lost.idxmax()

print(f"\n3. หลัง dropna() จะเหลือข้อมูล {len(df_drop)} แถว จาก {len(df)} แถว (คิดเป็น {remain_pct:.2f}%)")
print(f"   เมืองที่สูญเสียข้อมูลมากที่สุดคือ '{worst_city}' สูญเสียไป {rows_lost[worst_city]} แถว")
````

### โจทย์ 8.2
เติมค่าว่างของ `temperature_c` ด้วยสามวิธี ได้แก่ ค่าเฉลี่ยรวมทั้งตาราง ค่าเฉลี่ยของเมืองตนเอง และค่าของวันก่อนหน้าภายในเมืองเดียวกัน
จากนั้นเปรียบเทียบผลกระทบต่อค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานรายเมืองในรูปตารางเดียว

**อธิบาย:** วิธีใดเหมาะสมที่สุดกับข้อมูลชุดนี้ และวิธีใดทำให้ความแปรปรวนของข้อมูลลดลงอย่างผิดธรรมชาติ

````python
# คำตอบข้อ 8.2
# เติมค่าว่าง temperature_c ด้วย 3 วิธี
fill_global = df["temperature_c"].fillna(df["temperature_c"].mean())
fill_city = df["temperature_c"].fillna(df.groupby("city")["temperature_c"].transform("mean"))
fill_ffill = (df.sort_values(["city", "date"])
                .groupby("city")["temperature_c"]
                .ffill()
                .bfill())

comp_table = pd.DataFrame({
    "Mean_Orig": df.groupby("city")["temperature_c"].mean(),
    "Std_Orig": df.groupby("city")["temperature_c"].std(),
    "Mean_Global": df.assign(t=fill_global).groupby("city")["t"].mean(),
    "Std_Global": df.assign(t=fill_global).groupby("city")["t"].std(),
    "Mean_City": df.assign(t=fill_city).groupby("city")["t"].mean(),
    "Std_City": df.assign(t=fill_city).groupby("city")["t"].std(),
    "Mean_FFill": df.assign(t=fill_ffill).groupby("city")["t"].mean(),
    "Std_FFill": df.assign(t=fill_ffill).groupby("city")["t"].std(),
}).round(2)

print("ตารางเปรียบเทียบผลกระทบต่อค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานรายเมือง:")
print(comp_table.head(6))

print("\n[อธิบาย]:")
print("1. วิธีที่เหมาะสมที่สุดกับข้อมูลชุดนี้: คือ 'การเติมด้วยค่าเฉลี่ยของเมืองตนเอง' หรือ 'ค่าของวันก่อนหน้า (ffill)' เพราะสภาพอากาศมีความจำเพาะกับท้องถิ่น การใช้ข้อมูลในเมืองเดียวกันช่วยสะท้อนอุณหภูมิที่สมจริงที่สุด")
print("2. วิธีที่ทำให้ความแปรปรวนของข้อมูลลดลงอย่างผิดธรรมชาติ: คือ 'การเติมด้วยค่าเฉลี่ย (Mean Imputation)' โดยเฉพาะ 'ค่าเฉลี่ยรวมทั้งตาราง' เพราะการนำค่าคงที่ค่าเดียวมาเติมซ้ำๆ จะดึงอุณหภูมิของทุกเมืองเข้าหาค่ากลาง ทำให้ความแปรปรวน (Variance/Std) หดตัวลงอย่างผิดธรรมชาติ")
````

### โจทย์ 9.2

## ข้อ 10 `transform()` และ `filter()`

`agg()` ให้ผลลัพธ์หนึ่งแถวต่อหนึ่งกลุ่ม ในขณะที่ `transform()` คืนผลลัพธ์ที่มีจำนวนแถวเท่าเดิม
ส่วน `filter()` ใช้คัดเลือกทั้งกลุ่มตามเงื่อนไขระดับกลุ่ม

### ตัวอย่าง

````python
# transform คืนค่าที่มีจำนวนแถวเท่าตารางเดิม จึงนำไปสร้างคอลัมน์ใหม่ได้ทันที
df["city_avg_temp"] = df.groupby("city")["temperature_c"].transform("mean")
df["temp_diff"] = df["temperature_c"] - df["city_avg_temp"]
print(df[["city", "temperature_c", "city_avg_temp", "temp_diff"]].head(3).round(2))

# เปรียบเทียบรูปร่างของผลลัพธ์ระหว่าง agg และ transform
print(df.groupby("city")["temperature_c"].mean().shape,
      df.groupby("city")["temperature_c"].transform("mean").shape)
````

````python
# transform ร่วมกับฟังก์ชันที่นิยามเอง เช่น การคำนวณ z-score ภายในกลุ่ม
z = df.groupby("city")["temperature_c"].transform(lambda s: (s - s.mean()) / s.std())
print(z.describe().round(3))
````

````python
# filter คัดเลือกทั้งกลุ่มตามเงื่อนไขที่คำนวณจากกลุ่มนั้น
big_cities = df.groupby("city").filter(lambda g: g["temperature_c"].notna().sum() >= 88)
print("จำนวนเมืองก่อนคัด:", df["city"].nunique(), " หลังคัด:", big_cities["city"].nunique())
````

### โจทย์ 10.1
สร้างคอลัมน์ใหม่ใน `df` โดยไม่เปลี่ยนจำนวนแถว ได้แก่

- `rain_share` สัดส่วนปริมาณฝนของวันนั้นเทียบกับปริมาณฝนรวมของเมืองนั้น
- `temp_z` ค่ามาตรฐานของอุณหภูมิภายในเมืองตนเอง
- `city_rank_month` อันดับของเดือนตามอุณหภูมิเฉลี่ยภายในเมืองตนเอง

จากนั้นตรวจสอบด้วยโปรแกรมว่าผลรวมของ `rain_share` ในแต่ละเมืองมีค่าเท่ากับ 1

````python
# คำตอบข้อ 10.1
# สร้างคอลัมน์ใหม่โดยไม่เปลี่ยนจำนวนแถว
df["rain_share"] = df["rainfall_mm"] / df.groupby("city")["rainfall_mm"].transform("sum")
df["temp_z"] = df.groupby("city")["temperature_c"].transform(lambda s: (s - s.mean()) / s.std())

# อันดับของเดือนตามอุณหภูมิเฉลี่ยภายในเมืองตนเอง (อุณหภูมิเฉลี่ยสูงได้อันดับ 1)
month_avg = df.groupby(["city", "month"])["temperature_c"].mean().reset_index()
month_avg["city_rank_month"] = month_avg.groupby("city")["temperature_c"].rank(ascending=False, method="dense").astype(int)
df = df.merge(month_avg[["city", "month", "city_rank_month"]], on=["city", "month"], how="left")

# ตรวจสอบว่าผลรวมของ rain_share ในแต่ละเมืองเท่ากับ 1
rain_share_sum = df.groupby("city")["rain_share"].sum()
is_sum_one = np.allclose(rain_share_sum, 1.0)
print(f"ผลการตรวจสอบ: ผลรวมของ rain_share ในแต่ละเมืองเท่ากับ 1 หรือไม่: {is_sum_one}")

print("\nตัวอย่าง 5 แถวแรกของคอลัมน์ใหม่:")
print(df[["city", "date", "rainfall_mm", "rain_share", "temperature_c", "temp_z", "city_rank_month"]].head())
````

### โจทย์ 10.2
ใช้ `groupby().filter()` คัดเลือกเฉพาะเมืองที่มีข้อมูลอุณหภูมิไม่ว่างอย่างน้อย 88 วัน แล้วระบุว่าเมืองใดถูกคัดออก
จากนั้นเขียนโปรแกรมที่ให้ผลลัพธ์เดียวกันโดยไม่ใช้ `filter()` และเปรียบเทียบความกระชับของทั้งสองวิธี

````python
# คำตอบข้อ 10.2
# วิธีที่ 1: ใช้ groupby().filter()
df_filt1 = df.groupby("city").filter(lambda g: g["temperature_c"].notna().sum() >= 88)
dropped_cities = set(df["city"].unique()) - set(df_filt1["city"].unique())
print(f"วิธีที่ 1 (ใช้ filter): เมืองที่ถูกคัดออกคือ {dropped_cities}")

# วิธีที่ 2: ไม่ใช้ filter() (ใช้ boolean indexing ด้วย .isin())
valid_cities = df.groupby("city")["temperature_c"].count().loc[lambda s: s >= 88].index
df_filt2 = df[df["city"].isin(valid_cities)]

print(f"วิธีที่ 2 (ไม่ใช้ filter): ให้ผลลัพธ์เท่ากับวิธีที่ 1 หรือไม่: {df_filt1.equals(df_filt2)}")

print("\n[เปรียบเทียบความกระชับ]:")
print("- วิธี groupby().filter(): เขียนได้สั้นกระชับในบรรทัดเดียว เป็นรูปแบบ Declarative ที่เข้าใจง่าย แต่ทำงานช้ากว่าเพราะต้องวนลูปฟังก์ชันกับทุกกลุ่มในระดับ Python")
print("- วิธีที่ไม่ใช้ filter(): เขียนแยก 2 ขั้นตอน (หาเมืองที่ผ่านเกณฑ์แล้วนำมากรองด้วย .isin()) แต่ทำงานแบบ Vectorized เต็มรูปแบบ ทำให้ประมวลผลได้เร็วกว่ามากเมื่อชุดข้อมูลมีขนาดใหญ่")
````

### โจทย์ 10.3
หา 5 วันที่มีค่ามาตรฐานของอุณหภูมิสูงสุดของทั้งตาราง โดยคิดเทียบกับเมืองของตนเอง
ผลลัพธ์ต้องประกอบด้วย `city`, `date`, `temperature_c`, `temp_z`

**อธิบาย:** เหตุใดการหาค่าผิดปกติโดยเทียบภายในกลุ่มจึงเหมาะสมกับข้อมูลชุดนี้มากกว่าการใช้เกณฑ์เดียวกันทั้งตาราง

### โจทย์ 11.1
หาสามวันที่มีอุณหภูมิสูงสุดของแต่ละภูมิภาค โดยผลลัพธ์ต้องมีไม่เกิน 18 แถว
ให้ทำสองวิธี ได้แก่ การเรียงลำดับแล้วใช้ `groupby().head()` และการใช้ `groupby().apply()` ร่วมกับ `nlargest`
จากนั้นวัดเวลาการทำงานเปรียบเทียบ

````python
# คำตอบข้อ 11.1
import time

# วิธีที่ 1: การเรียงลำดับแล้วใช้ groupby().head()
t0 = time.perf_counter()
res_m1 = (df.sort_values(["region", "temperature_c"], ascending=[True, False])
            .groupby("region")
            .head(3)
            [["region", "city", "date", "temperature_c"]]
            .reset_index(drop=True))
t_m1 = time.perf_counter() - t0

# วิธีที่ 2: การใช้ groupby().apply() ร่วมกับ nlargest
t0 = time.perf_counter()
res_m2 = (df.groupby("region")
            .apply(lambda g: g.nlargest(3, "temperature_c"), include_groups=False)
            .reset_index()
            [["region", "city", "date", "temperature_c"]])
t_m2 = time.perf_counter() - t0

print(f"จำนวนแถวของผลลัพธ์: {len(res_m1)} แถว (ไม่เกิน 18 แถว)")
print(f"เวลาทำงานวิธีที่ 1 (sort + head): {t_m1*1000:.3f} ms")
print(f"เวลาทำงานวิธีที่ 2 (groupby.apply + nlargest): {t_m2*1000:.3f} ms")
print(f"สรุป: วิธีที่ 1 เร็วกว่าวิธีที่ 2 ประมาณ {t_m2 / t_m1:.1f} เท่า")
print("\nตัวอย่างผลลัพธ์:")
print(res_m1.head(6))
````

ทดลองใช้ `method` ทั้งสามแบบตามตัวอย่าง แล้ว **อธิบาย** ความแตกต่างของผลลัพธ์เมื่อมีค่าฝนเท่ากันจำนวนมาก

### โจทย์ 12.2
สร้าง `pd.crosstab` ระหว่าง `region` กับช่วงอุณหภูมิที่นักศึกษากำหนดเอง จำนวน 3 แบบ ได้แก่ จำนวนนับ สัดส่วนตามแถว และสัดส่วนตามคอลัมน์

````python
# คำตอบข้อ 12.2
temp_bins = pd.cut(
    df["temperature_c"],
    bins=[-100, 15, 25, 35, 100],
    labels=["เย็น (<15)", "อบอุ่น (15-25)", "ร้อน (25-35)", "ร้อนจัด (>35)"]
)

# 1. จำนวนนับ
ct_count = pd.crosstab(df["region"], temp_bins)
print("1. Crosstab: จำนวนนับ (Counts):")
print(ct_count)

# 2. สัดส่วนตามแถว
ct_row = pd.crosstab(df["region"], temp_bins, normalize="index").round(3)
print("\n2. Crosstab: สัดส่วนตามแถว (Row proportions):")
print(ct_row)

# 3. สัดส่วนตามคอลัมน์
ct_col = pd.crosstab(df["region"], temp_bins, normalize="columns").round(3)
print("\n3. Crosstab: สัดส่วนตามคอลัมน์ (Column proportions):")
print(ct_col)
````

### โจทย์ 12.3
เปรียบเทียบผลของ `pd.cut()` กับ `pd.qcut()` เมื่อแบ่งอุณหภูมิออกเป็นสี่กลุ่ม
**อธิบาย:** เหตุใดจำนวนสมาชิกในแต่ละกลุ่มจึงแตกต่างกัน และควรเลือกใช้วิธีใดในสถานการณ์ใด

````python
# คำตอบข้อ 12.3
cut_4 = pd.cut(df["temperature_c"], bins=4)
qcut_4 = pd.qcut(df["temperature_c"], q=4)

print("จำนวนสมาชิกในแต่ละกลุ่มของ pd.cut():")
print(cut_4.value_counts().sort_index())

print("\nจำนวนสมาชิกในแต่ละกลุ่มของ pd.qcut():")
print(qcut_4.value_counts().sort_index())

print("\n[อธิบาย]:")
print("1. เหตุใดจำนวนสมาชิกในแต่ละกลุ่มจึงแตกต่างกัน:")
print("   - pd.cut() แบ่งตาม 'ความกว้างช่วงเท่ากัน (Equal-width)': เมื่อมีค่า Outlier (เช่น 999 °C) จะทำให้ช่วงกว้างมาก และข้อมูลส่วนใหญ่เทไปกองอยู่ในช่วงแรกเกือบหมด ขณะที่ช่วงอื่นแทบไม่มีข้อมูลเลย")
print("   - pd.qcut() แบ่งตาม 'เปอร์เซ็นไทล์/ความถี่เท่ากัน (Equal-frequency)': จะจัดให้แต่ละกลุ่มมีจำนวนแถวใกล้เคียงกันเสมอ (ประมาณ 25% ต่อกลุ่ม) โดยขนาดความกว้างของช่วงจะไม่เท่ากัน")
print("2. การเลือกใช้ในสถานการณ์จริง:")
print("   - ควรใช้ pd.cut() เมื่อต้องการแบ่งตามเกณฑ์มาตรฐานทางกายภาพหรือธุรกิจที่กำหนดไว้แน่นอน (เช่น ช่วงอุณหภูมิทางการแพทย์ หรือเกณฑ์ตัดเกรด)")
print("   - ควรใช้ pd.qcut() เมื่อต้องการจัดกลุ่มทางสถิติให้แต่ละกลุ่มมีจำนวนตัวอย่างสมดุลกัน (เช่น การแบ่งกลุ่มลูกค้าเป็น 4 Quartile)")
````

````python
# คำตอบข้อ 13.2
unstack_rain = df.groupby(["city", "month"])["rainfall_mm"].sum().unstack()

# เพิ่มยอดรวมท้ายแถวและท้ายคอลัมน์
unstack_rain["รวม"] = unstack_rain.sum(axis=1)
unstack_rain.loc["รวม"] = unstack_rain.sum(axis=0)

is_same = np.allclose(pivot_rain.values, unstack_rain.values)
print(f"ผลการตรวจสอบด้วย np.allclose: ให้ผลตรงกันหรือไม่: {is_same}")

print("\n[อธิบาย]:")
print("ควรเลือกใช้ pivot_table() หรือ groupby().unstack() ในสถานการณ์ใด:")
print("- ควรเลือกใช้ pivot_table(): เมื่อต้องการสรุปผลข้อมูลสองมิติเพื่อการนำเสนอหรือรายงาน เพราะมีพารามิเตอร์ margins=True ช่วยคำนวณยอดรวมได้ทันทีในคำสั่งเดียว")
print("- ควรเลือกใช้ groupby().unstack(): เมื่อต้องการเขียนเป็น Pipeline ร่วมกับการประมวลผลข้อมูลหลายขั้นตอน มีความยืดหยุ่นสูงในการคำนวณร่วมกับ GroupBy ขั้นสูง และมีประสิทธิภาพดีกว่าเมื่อไม่ต้องการผลรวม margins")
````

แปลงตารางรูปกว้างจากโจทย์ 13.1 กลับเป็นรูปยาวด้วย `.melt()` แล้วตรวจสอบว่าข้อมูลครบถ้วนเท่าเดิม
**อธิบาย:** ข้อมูลรูปยาวเหมาะสมกับการจัดเก็บอย่างไร และข้อมูลรูปกว้างเหมาะสมกับการนำเสนออย่างไร

````python
# คำตอบข้อ 13.3
# ตัดแถวและคอลัมน์ 'รวม' ออกก่อนแปลงกลับเป็นรูปยาว
wide_tab = pivot_rain.drop(index="รวม", columns="รวม").reset_index()
long_tab = wide_tab.melt(id_vars="city", var_name="month", value_name="rainfall_mm")

expected_count = df["city"].nunique() * df["month"].nunique()
print(f"จำนวนแถวของตารางรูปยาว: {len(long_tab)} แถว (ครบถ้วน {expected_count} แถว: {len(long_tab) == expected_count})")
print(f"ผลรวมปริมาณฝนตรงกับข้อมูลเดิมหรือไม่: {np.isclose(long_tab['rainfall_mm'].sum(), df['rainfall_mm'].sum())}")
print("\nตัวอย่าง 5 แถวแรกของตารางรูปยาว:")
print(long_tab.head())

print("\n[อธิบาย]:")
print("- ข้อมูลรูปยาว (Long format): เหมาะสมสำหรับการจัดเก็บในฐานข้อมูล (Database/Data Warehouse) และการนำไปวิเคราะห์ด้วยโมเดลหรือพล็อตกราฟ เพราะแต่ละแถวคือ 1 observation ทำให้ขยายมิติข้อมูลใหม่ได้ง่ายโดยไม่ต้องแก้ Schema")
print("- ข้อมูลรูปกว้าง (Wide format): เหมาะสมสำหรับการนำเสนอผลลัพธ์ (Reporting / Dashboard) ให้มนุษย์อ่านและเปรียบเทียบข้อมูลระหว่างกลุ่มหรือข้ามช่วงเวลาได้ง่ายในหน้าจอเดียว")
````

### โจทย์ 13.4
สร้างตารางที่มี index เป็นคู่ `(region, city)` และคอลัมน์เป็นคู่ `(month, metric)` โดย metric ประกอบด้วย `temperature_c` และ `rainfall_mm`
จากนั้นใช้ `.swaplevel()` และ `.sort_index()` จัดลำดับคอลัมน์ใหม่ และเลือกข้อมูลเฉพาะเดือนที่ 2 ของทุกเมืองในภูมิภาคเอเชียด้วย `.loc[]`

ต้องเรียก `.sort_index()` ทุกครั้งหลัง `swaplevel` เพราะการสลับชั้นทำให้ลำดับ index ไม่เรียง และการเลือกข้อมูลแบบช่วงบน index ที่ไม่เรียงจะทำงานไม่ถูกต้องหรือเกิด `UnsortedIndexError`

````python
# คำตอบข้อ 13.4
# สร้างตาราง index เป็น (region, city) และคอลัมน์เป็น (month, metric)
pivot_134 = df.pivot_table(
    index=["region", "city"],
    columns="month",
    values=["temperature_c", "rainfall_mm"],
    aggfunc="mean"
)

# สลับลำดับชั้นคอลัมน์ให้ month อยู่ชั้นนอก และ metric อยู่ชั้นใน พร้อมเรียงลำดับ index
pivot_swapped = pivot_134.swaplevel(0, 1, axis=1).sort_index(axis=1)

# เลือกข้อมูลเฉพาะเดือนที่ 2 ของทุกเมืองในภูมิภาคเอเชียด้วย .loc[]
asia_m2 = pivot_swapped.loc["Asia", 2]

print("ข้อมูลเฉพาะเดือนที่ 2 ของทุกเมืองในภูมิภาคเอเชีย:")
print(asia_m2.round(2))
````

2. ปริมาณฝนรวมทั้งตารางที่เปลี่ยนแปลงไป คิดเป็นร้อยละ
3. ผลของการกำหนด `validate="many_to_one"` ในกรณีนี้

**อธิบาย:** เหตุใดข้อผิดพลาดลักษณะนี้จึงอันตรายกว่าการรวมตารางที่ล้มเหลวจนเกิดข้อความแจ้งข้อผิดพลาด

````python
# คำตอบข้อ 14.2
# เพิ่มแถวซ้ำของเมือง Bangkok ใน city_info
city_info_dup = pd.concat([city_info, city_info[city_info["city"] == "Bangkok"]], ignore_index=True)

# รวมตารางใหม่
merged_dup = df.merge(city_info_dup.drop(columns=["country", "region"]), on="city", how="left")

# รายงานผล
added_rows = len(merged_dup) - len(df)
rain_change_pct = ((merged_dup["rainfall_mm"].sum() - df["rainfall_mm"].sum()) / df["rainfall_mm"].sum()) * 100

print(f"1. จำนวนแถวที่เพิ่มขึ้น: {added_rows} แถว")
print(f"2. ปริมาณฝนรวมทั้งตารางที่เปลี่ยนแปลงไป: +{rain_change_pct:.2f}%")

print("\n3. ผลของการกำหนด validate='many_to_one':")
try:
    df.merge(city_info_dup.drop(columns=["country", "region"]), on="city", how="left", validate="many_to_one")
except Exception as e:
    print(f"   เกิดข้อผิดพลาดตามที่คาดไว้: {type(e).__name__} -> {e}")

print("\n[อธิบาย]:")
print("เหตุใดข้อผิดพลาดลักษณะนี้จึงอันตรายกว่าการรวมตารางที่ล้มเหลวจนเกิด Error:")
print("เพราะโปรแกรมทำงานได้จนจบโดยไม่มีข้อความ Error แจ้งเตือน (Silent failure) ส่งผลให้เกิดปัญหา Fan-out (ข้อมูลถูกคูณจำนวนแถว) ทำให้ตัวเลขสรุปเชิงสถิติ เช่น ยอดขาย หรือปริมาณฝนรวม บวมขึ้นเกินจริงอย่างมหาศาลโดยไม่มีใครรู้ตัว ซึ่งนำไปสู่การตัดสินใจที่ผิดพลาดอย่างร้ายแรง")
````

````python
# คำตอบข้อ 15.2
# สร้าง DataFrame ใหม่ที่มีคอลัมน์ไม่ครบ และมีคอลัมน์เกินมาหนึ่งคอลัมน์
df_sub = df.iloc[:10, :5].copy()
df_sub["special_note"] = "บันทึกพิเศษ"

concat_out = pd.concat([df, df_sub], join="outer", ignore_index=True)
concat_inn = pd.concat([df, df_sub], join="inner", ignore_index=True)

print(f"ขนาดตารางเมื่อใช้ join='outer': {concat_out.shape} (คอลัมน์แบบ Union = {len(concat_out.columns)} คอลัมน์)")
print(f"ขนาดตารางเมื่อใช้ join='inner': {concat_inn.shape} (คอลัมน์แบบ Intersection = {len(concat_inn.columns)} คอลัมน์)")

print("\n[อธิบาย]:")
print("- ควรเลือกใช้ join='outer': เมื่อต้องการเก็บรักษาข้อมูลทุกคอลัมน์ของทุกตารางไว้ทั้งหมด โดยยอมให้มีค่าว่าง (NaN) ในแถวที่ไม่มีข้อมูลคอลัมน์นั้น เพื่อไม่ให้ข้อมูลสูญหาย")
print("- ควรเลือกใช้ join='inner': เมื่อต้องการเฉพาะคอลัมน์ที่มีร่วมกันในทุกชุดข้อมูลอย่างสมบูรณ์ และไม่ต้องการให้เกิดค่าว่าง (NaN) ในตารางผลลัพธ์")
````

````python
# np.select เขียนเงื่อนไขหลายชั้นแบบ vectorized ซึ่งเร็วกว่า apply แบบ axis=1
conditions = [
    (df["temperature_c"] > 30) & (df["humidity_pct"] > 70),
    (df["rainfall_mm"] > 10),
    (df["wind_speed_kmh"] > 25),
]
choices = ["ร้อนชื้น", "ฝนหนัก", "ลมแรง"]
df["weather_flag"] = np.select(conditions, choices, default="ปกติ")
print(df["weather_flag"].value_counts())
````

````python
# pipe ส่ง DataFrame เข้าเป็น argument แรกของฟังก์ชัน จึงต่อขั้นตอนหลายขั้นเป็นลำดับเดียวได้
# แต่ละฟังก์ชันต้องรับ DataFrame และคืน DataFrame เพื่อให้ส่งต่อขั้นถัดไปได้

def clean_range(data, col, low, high):
    return data.assign(**{col: data[col].where(data[col].between(low, high))})

def add_flag(data):
    conditions = [
        (data["temperature_c"] > 30) & (data["humidity_pct"] > 70),
        data["rainfall_mm"] > 10,
    ]
    return data.assign(weather_flag=np.select(conditions, ["ร้อนชื้น", "ฝนหนัก"], default="ปกติ"))

def summarize(data, by="region"):
    return (data.groupby(by)
                .agg(avg_temp=("temperature_c", "mean"),
                     heavy_rain_days=("rainfall_mm", lambda s: (s > 10).sum()))
                .round(2))

result = (df
          .pipe(clean_range, col="temperature_c", low=-60, high=60)
          .pipe(clean_range, col="humidity_pct", low=0, high=100)
          .pipe(add_flag)
          .pipe(summarize, by="region"))
print(result)
````

````python
# ข้อจำกัดของการเขียนแบบลำดับคือไม่มีตัวแปรกลางให้ตรวจสอบ
# วิธีแก้คือแทรกฟังก์ชันที่ตรวจสภาพข้อมูลแล้วคืนค่าเดิมกลับไป

def trace(data, label):
    print(f"[{label}] จำนวนแถว = {len(data)}  ค่าว่างของอุณหภูมิ = {data['temperature_c'].isna().sum()}")
    return data

out = (df
       .pipe(clean_range, col="temperature_c", low=-60, high=60)
       .pipe(trace, "หลังทำความสะอาดอุณหภูมิ")
       .pipe(add_flag)
       .pipe(trace, "หลังเพิ่มคอลัมน์ flag"))
print(out)
````

### โจทย์ 16.1
สร้างคอลัมน์ `weather_flag` ของตนเองจากเงื่อนไขที่ใช้อย่างน้อยสามคอลัมน์ร่วมกัน และมีอย่างน้อยสี่กลุ่ม
โดยเขียนสามวิธี ได้แก่ `df.apply(..., axis=1)`, `np.select` และ `np.where` ซ้อนกัน
จากนั้นตรวจสอบว่าทั้งสามวิธีให้ผลตรงกันทุกแถว และรายงานว่าวิธีที่เร็วที่สุดเร็วกว่าวิธีที่ช้าที่สุดกี่เท่า

````python
# คำตอบข้อ 16.1
import time

# กำหนดเงื่อนไขจาก 3 คอลัมน์ (temperature_c, humidity_pct, rainfall_mm) และแบ่งเป็น 4 กลุ่ม
# กลุ่ม 1: ร้อนชื้นมาก (temp > 30, humid > 70, rain > 5)
# กลุ่ม 2: ร้อนแห้ง (temp > 30, humid <= 70)
# กลุ่ม 3: หนาวชื้น (temp < 15, rain > 5)
# กลุ่ม 4: ปกติ (อื่นๆ)

# วิธีที่ 1: df.apply(..., axis=1)
def classify_weather(row):
    t, h, r = row["temperature_c"], row["humidity_pct"], row["rainfall_mm"]
    if t > 30 and h > 70 and r > 5:
        return "ร้อนชื้นมาก"
    elif t > 30 and h <= 70:
        return "ร้อนแห้ง"
    elif t < 15 and r > 5:
        return "หนาวชื้น"
    return "ปกติ"

t0 = time.perf_counter()
f_apply = df.apply(classify_weather, axis=1)
t_apply = time.perf_counter() - t0

# วิธีที่ 2: np.select
conds = [
    (df["temperature_c"] > 30) & (df["humidity_pct"] > 70) & (df["rainfall_mm"] > 5),
    (df["temperature_c"] > 30) & (df["humidity_pct"] <= 70),
    (df["temperature_c"] < 15) & (df["rainfall_mm"] > 5),
]
choices = ["ร้อนชื้นมาก", "ร้อนแห้ง", "หนาวชื้น"]

t0 = time.perf_counter()
f_select = pd.Series(np.select(conds, choices, default="ปกติ"), index=df.index)
t_select = time.perf_counter() - t0

# วิธีที่ 3: np.where ซ้อนกัน
t0 = time.perf_counter()
f_where = pd.Series(
    np.where(
        conds[0], choices[0],
        np.where(conds[1], choices[1],
                 np.where(conds[2], choices[2], "ปกติ"))
    ),
    index=df.index
)
t_where = time.perf_counter() - t0

# ตรวจสอบความถูกต้อง
is_all_match = (f_apply == f_select).all() and (f_select == f_where).all()
print(f"ผลการตรวจสอบ: ทั้งสามวิธีให้ผลลัพธ์ตรงกันทุกแถวหรือไม่: {is_all_match}")

# รายงานผลความเร็ว
perf = {"df.apply(axis=1)": t_apply, "np.select": t_select, "np.where": t_where}
fastest = min(perf, key=perf.get)
slowest = max(perf, key=perf.get)
speedup = perf[slowest] / perf[fastest]

print("\nเวลาในการประมวลผล:")
for name, elapsed in perf.items():
    print(f" - {name}: {elapsed*1000:.3f} ms")
print(f"\nรายงาน: วิธีที่เร็วที่สุด ({fastest}) เร็วกว่าวิธีที่ช้าที่สุด ({slowest}) ถึง {speedup:.1f} เท่า")

df["weather_flag"] = f_select
print("\nความถี่ของแต่ละกลุ่มใน weather_flag:")
print(df["weather_flag"].value_counts())
````

````python
# คำตอบข้อ 16.2
region_map = {
    "Asia": "AS",
    "Europe": "EU",
    "Africa": "AF",
    "North America": "NA",
    "South America": "SA",
    "Oceania": "OC"
}

# สร้างคอลัมน์ region_code ด้วย .map()
df["region_code"] = df["region"].map(region_map)

print("ตัวอย่างผลลัพธ์ของคอลัมน์ region_code:")
print(df[["region", "region_code"]].drop_duplicates())

print("\n[อธิบาย]:")
print("1. .apply() บน Series แตกต่างจาก .map() อย่างไร:")
print("   - .map() ออกแบบมาเพื่อการแปลงค่าแบบ Element-wise โดยตรง สามารถรับ Input เป็น Dictionary, Series หรือ Function ก็ได้ โดยหากค่าใดไม่มีอยู่ใน dict จะถูกแทนด้วย NaN ทันที")
print("   - .apply() รับ Input ได้เฉพาะฟังก์ชัน (Callable) เท่านั้น ไม่สามารถส่ง dictionary เข้าไปตรงๆ ได้ เหมาะกับฟังก์ชันที่มีตรรกะซับซ้อน")
print("2. วิธีจัดการค่าที่ไม่มีอยู่ใน dict เมื่อใช้ .map():")
print("   - วิธีที่ 1: ใช้ .fillna() ต่อท้าย เช่น df['region'].map(region_map).fillna('UNKNOWN')")
print("   - วิธีที่ 2: ใช้ dict.get() ผ่าน lambda เช่น df['region'].map(lambda x: region_map.get(x, 'UNKNOWN'))")
````

### โจทย์ 16.3
เขียนฟังก์ชัน `heat_index(temp, humidity)` ตามสูตรอย่างง่ายที่นักศึกษากำหนดเอง แล้วสร้างคอลัมน์ใหม่สองวิธี
ได้แก่ การเรียกผ่าน `apply(..., axis=1)` และการส่ง Series ทั้งคอลัมน์เข้าฟังก์ชันโดยตรง
**อธิบาย:** เหตุใดวิธีที่สองจึงทำงานได้ทั้งที่ฟังก์ชันเขียนเหมือนกัน และเงื่อนไขใดที่ทำให้วิธีที่สองใช้ไม่ได้

````python
# คำตอบข้อ 16.3
# กำหนดสูตรคำนวณ heat_index อย่างง่าย
def heat_index(temp, humidity):
    return temp + 0.1 * humidity

# วิธีที่ 1: เรียกผ่าน apply(..., axis=1)
df["heat_idx_apply"] = df.apply(lambda r: heat_index(r["temperature_c"], r["humidity_pct"]), axis=1)

# วิธีที่ 2: ส่ง Series ทั้งคอลัมน์เข้าฟังก์ชันโดยตรง
df["heat_idx_vec"] = heat_index(df["temperature_c"], df["humidity_pct"])

is_same_result = np.allclose(df["heat_idx_apply"].dropna(), df["heat_idx_vec"].dropna())
print(f"ผลการตรวจสอบ: ทั้งสองวิธีให้ผลตรงกันหรือไม่: {is_same_result}")

print("\n[อธิบาย]:")
print("1. เหตุใดวิธีที่สองจึงทำงานได้ทั้งที่ฟังก์ชันเขียนเหมือนกัน:")
print("   เพราะตัวดำเนินการ +, * ในภาษา Python ถูก Overload โดย pandas Series และ NumPy ให้ทำงานแบบ Vectorization / Element-wise โดยอัตโนมัติ เมื่อส่ง Series เข้าไป ฟังก์ชันจึงประมวลผลพร้อมกันทั้งคอลัมน์ได้อย่างมีประสิทธิภาพ")
print("2. เงื่อนไขที่ทำให้วิธีที่สองใช้ไม่ได้:")
print("   เมื่อภายในฟังก์ชันมีตรรกะแบบ Scalar เช่น การใช้คำสั่ง if-else, while loop หรือฟังก์ชันจากไลบรารี math (เช่น math.sqrt, math.sin) เพราะคำสั่ง if ไม่สามารถตัดสินใจบน Series ที่มีหลายค่าพร้อมกันได้ (จะเกิด ValueError: The truth value of a Series is ambiguous)")
````

### โจทย์ 17.3
สร้างกราฟแท่งซ้อนแสดงจำนวนวันในแต่ละช่วงปริมาณฝน โดยแบ่งช่วงด้วย `pd.cut()` และจำแนกตามภูมิภาค
จากนั้นสร้างกราฟแบบ `stacked=False` เปรียบเทียบ
**อธิบาย:** กราฟแท่งซ้อนเหมาะสมกว่าในกรณีใด และกราฟแท่งเรียงข้างกันเหมาะสมกว่าในกรณีใด

````python
# คำตอบข้อ 17.3
# แบ่งช่วงปริมาณฝนด้วย pd.cut()
rain_bins = pd.cut(
    df["rainfall_mm"],
    bins=[-1, 0, 5, 15, 100],
    labels=["ไม่มีฝน (0 mm)", "ฝนตกเล็กน้อย (0-5 mm)", "ฝนปานกลาง (5-15 mm)", "ฝนตกหนัก (>15 mm)"]
)

crosstab_rain = pd.crosstab(df["region"], rain_bins)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# กราฟแท่งซ้อน (stacked=True)
crosstab_rain.plot(kind="bar", stacked=True, ax=axes[0], colormap="viridis")
axes[0].set_title("กราฟแท่งซ้อนแสดงจำนวนวันตามช่วงปริมาณฝน (Stacked = True)")
axes[0].set_ylabel("จำนวนวัน (Days)")
axes[0].tick_params(axis="x", rotation=30)

# กราฟแท่งเรียงข้างกัน (stacked=False)
crosstab_rain.plot(kind="bar", stacked=False, ax=axes[1], colormap="viridis")
axes[1].set_title("กราฟแท่งเรียงข้างกันเปรียบเทียบแต่ละช่วง (Stacked = False)")
axes[1].set_ylabel("จำนวนวัน (Days)")
axes[1].tick_params(axis="x", rotation=30)

plt.tight_layout()
plt.show()

print("[อธิบาย]:")
print("- กราฟแท่งซ้อน (stacked=True): เหมาะสมเมื่อต้องการดูภาพรวมยอดรวมทั้งหมด (Total / Aggregate) ของแต่ละภูมิภาค ควบคู่ไปกับการดูสัดส่วนองค์ประกอบภายใน")
print("- กราฟแท่งเรียงข้างกัน (stacked=False): เหมาะสมเมื่อต้องการเปรียบเทียบค่าความถี่ระหว่างแต่ละช่วงฝนข้ามภูมิภาคโดยตรง เพราะทุกแท่งเริ่มต้นจากฐานแกนเดียวกัน (Common baseline) ทำให้สายตาสามารถเปรียบเทียบความสูงของแท่งได้อย่างแม่นยำ")
````

## ข้อ 18 โจทย์ประยุกต์รวม

ให้ใช้ตารางที่รวมข้อมูลจากทั้ง 3 แหล่งและผ่านการทำความสะอาดแล้ว ตอบคำถามต่อไปนี้
แต่ละข้อต้องประกอบด้วย **ตารางหรือกราฟหนึ่งชิ้น** และ **ข้อสรุปเป็นข้อความสั้น ๆ**

**ข้อกำหนดเพิ่มเติม**

- ให้เขียนโปรแกรมในรูปแบบลำดับต่อเนื่องที่อ่านเข้าใจได้ ไม่ใช้ตัวแปรชั่วคราวจำนวนมากโดยไม่จำเป็น
- ห้ามใช้การวนซ้ำทีละแถว
- หากข้อสรุปขึ้นอยู่กับวิธีจัดการค่าว่างหรือค่าผิดปกติ ให้ระบุวิธีที่เลือกใช้ไว้ด้วย

1. ภูมิภาคใดมีสภาพอากาศแปรปรวนมากที่สุดในไตรมาสแรก เมื่อวัดจากส่วนเบี่ยงเบนมาตรฐานของอุณหภูมิรายวัน
   และความแปรปรวนดังกล่าวเกิดจากทุกเมืองในภูมิภาคนั้นเท่ากันหรือเกิดจากเมืองใดเมืองหนึ่งเป็นหลัก

````python
# คำตอบข้อ 18.1
# วิธีจัดการค่าผิดปกติ: แทนค่าอุณหภูมิที่อยู่นอกช่วง -60 ถึง 60 ด้วย NaN ก่อนคำนวณ std เพื่อไม่ให้ค่า 999 °C บิดเบือนความแปรปรวน
clean_data = (
    df.assign(
        temperature_c=df["temperature_c"].where(df["temperature_c"].between(-60, 60)),
        humidity_pct=df["humidity_pct"].where(df["humidity_pct"].between(0, 100)),
    )
    .merge(city_info.drop(columns=["country", "region"]), on="city", how="left")
    .merge(region_info, on="region", how="left")
)

# คำนวณส่วนเบี่ยงเบนมาตรฐานระดับภูมิภาค และระดับเมือง
region_std = clean_data.groupby("region")["temperature_c"].std().sort_values(ascending=False)
city_std_in_region = clean_data.groupby(["region", "city"])["temperature_c"].std()

# ชิ้นงาน: ตารางเปรียบเทียบส่วนเบี่ยงเบนมาตรฐาน
std_report = pd.DataFrame({
    "Region_Overall_Std": region_std,
    "Min_City_Std": clean_data.groupby("region")["temperature_c"].apply(lambda g: clean_data.loc[g.index].groupby("city")["temperature_c"].std().min()),
    "Max_City_Std": clean_data.groupby("region")["temperature_c"].apply(lambda g: clean_data.loc[g.index].groupby("city")["temperature_c"].std().max()),
}).round(2)

print("ตารางแสดงส่วนเบี่ยงเบนมาตรฐานของอุณหภูมิรายวัน:")
print(std_report)

top_reg = region_std.index[0]
print(f"\nค่า Std รายเมืองในภูมิภาค {top_reg}:")
print(clean_data[clean_data["region"] == top_reg].groupby("city")["temperature_c"].std().round(2))

print(f"""\n[ข้อสรุป]:
1. ภูมิภาคที่มีสภาพอากาศแปรปรวนมากที่สุดในไตรมาสแรกคือ '{top_reg}' (Std รวม = {region_std.max():.2f} °C)
2. ความแปรปรวนดังกล่าวไม่ได้เกิดจากเมืองใดเมืองหนึ่งผันผวนรุนแรงในแต่ละวัน เพราะทุกเมืองในเอเชียมี Std รายวันต่ำและใกล้เคียงกัน (ประมาณ 2.5 - 2.9 °C) 
   แต่เกิดจาก 'ความแตกต่างของสภาพภูมิอากาศระหว่างเมือง (Spatial Variation)' เนื่องจากภูมิภาคเอเชียมีทั้งเมืองเขตหนาว/อบอุ่น (Beijing, Tokyo อุณหภูมิเฉลี่ย ~15-16 °C) 
   และเมืองเขตร้อน (Bangkok, Singapore อุณหภูมิเฉลี่ย ~32 °C) เมื่อรวมข้อมูลระดับภูมิภาคจึงทำให้ส่วนเบี่ยงเบนมาตรฐานรวมสูงที่สุด""")
````

2. เมืองที่มีฝนตกหนักบ่อยที่สุด 5 อันดับแรกเมื่อวัดด้วยสัดส่วน ให้เปรียบเทียบกับผลที่ได้เมื่อวัดด้วยจำนวนวัน
   และอธิบายว่าเหตุใดจึงให้ผลต่างกัน

````python
# คำตอบข้อ 18.2
# คำนวณจำนวนวันฝนตกหนัก (>10 mm) และสัดส่วนต่อจำนวนวันที่มีข้อมูลจริง
heavy_cnt = (clean_data["rainfall_mm"] > 10).groupby(clean_data["city"]).sum()
valid_cnt = clean_data["rainfall_mm"].notna().groupby(clean_data["city"]).sum()
heavy_ratio = ((heavy_cnt / valid_cnt) * 100).round(2)

# ชิ้นงาน: ตารางเปรียบเทียบอันดับ
comp_top_rain = (
    pd.DataFrame({
        "จำนวนวันฝนตกหนัก (วัน)": heavy_cnt,
        "สัดส่วนวันฝนตกหนัก (%)": heavy_ratio,
        "อันดับตามจำนวนวัน": heavy_cnt.rank(ascending=False, method="min").astype(int),
        "อันดับตามสัดส่วน": heavy_ratio.rank(ascending=False, method="min").astype(int),
    })
    .sort_values("สัดส่วนวันฝนตกหนัก (%)", ascending=False)
)

print("ตารางเปรียบเทียบเมืองที่มีฝนตกหนักบ่อยที่สุด 5 อันดับแรก (เรียงตามสัดส่วน):")
print(comp_top_rain.head(5))

print("""\n[ข้อสรุป]:
1. เมืองที่มีฝนตกหนักบ่อยที่สุด 5 อันดับแรกตามสัดส่วน ได้แก่ Singapore (48.89%), Bangkok (26.44%), Lagos (25.00%), Sao Paulo (12.50%) และ Chiang Mai (10.59%)
2. เปรียบเทียบกับผลตามจำนวนวัน: อันดับ 1-3 เหมือนกัน แต่ Chiang Mai ไต่อันดับขึ้นมาติด Top 5 ในเกณฑ์สัดส่วน
3. เหตุผลที่ให้ผลต่างกัน: เพราะ Chiang Mai มีจำนวนวันที่มีข้อมูลบันทึกจริงน้อยกว่าเมืองอื่น (มี Missing values) การวัดด้วยจำนวนวันแบบสัมบูรณ์จึงทำให้เมืองที่มีข้อมูลสูญหายเสียเปรียบ ขณะที่การวัดด้วยสัดส่วนเป็นการปรับเทียบฐานข้อมูลจริงอย่างเป็นธรรม""")
````

3. จำนวนประชากรของเมืองมีความสัมพันธ์กับอุณหภูมิหรือปริมาณฝนหรือไม่
   ให้คำนวณค่าสหสัมพันธ์ประกอบ พร้อมระบุข้อควรระวังในการตีความอย่างน้อยหนึ่งประการ

````python
# คำตอบข้อ 18.3
# สรุปข้อมูลระดับเมือง (1 แถวต่อ 1 เมือง)
city_df_corr = (
    clean_data.groupby("city")
    .agg(
        population=("population_millions", "first"),
        avg_temp=("temperature_c", "mean"),
        total_rain=("rainfall_mm", "sum"),
    )
    .reset_index()
)

corr_matrix_city = city_df_corr[["population", "avg_temp", "total_rain"]].corr().round(3)

# ชิ้นงาน: ตารางสหสัมพันธ์และกราฟจุด
print("ตารางค่าสหสัมพันธ์ระดับเมือง (City-level Correlation):")
print(corr_matrix_city)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
city_df_corr.plot(kind="scatter", x="population", y="avg_temp", ax=axes[0], color="royalblue", s=60, edgecolors="black")
axes[0].set_title(f"Population vs. Avg Temp (r = {corr_matrix_city.loc['population', 'avg_temp']})")
axes[0].set_xlabel("Population (Millions)")
axes[0].set_ylabel("Average Temperature (°C)")
axes[0].grid(True, linestyle="--", alpha=0.5)

city_df_corr.plot(kind="scatter", x="population", y="total_rain", ax=axes[1], color="seagreen", s=60, edgecolors="black")
axes[1].set_title(f"Population vs. Total Rain (r = {corr_matrix_city.loc['population', 'total_rain']})")
axes[1].set_xlabel("Population (Millions)")
axes[1].set_ylabel("Total Rain (mm)")
axes[1].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print(f"""\n[ข้อสรุปและข้อควรระวัง]:
1. ข้อสรุป: จำนวนประชากรของเมืองแทบไม่มีความสัมพันธ์กับอุณหภูมิ (r = {corr_matrix_city.loc['population', 'avg_temp']}) หรือปริมาณฝน (r = {corr_matrix_city.loc['population', 'total_rain']}) ค่าความสัมพันธ์ใกล้ 0 แสดงว่าประชากรไม่ได้ถูกกำหนดด้วยสภาพอากาศในชุดข้อมูลนี้
2. ข้อควรระวังในการตีความ:
   - Correlation does not imply causation: สหสัมพันธ์ไม่ได้บอกเหตุและผล การตั้งถิ่นฐานและขนาดเมืองขึ้นกับปัจจัยทางประวัติศาสตร์ เศรษฐกิจ และชัยภูมิ
   - Confounding Variables: การวิเคราะห์ระดับเมืองมีปัจจัยแทรกซ้อน เช่น พื้นที่ของเขตเมือง (City land area) หรือตำแหน่งละติจูด
   - Influence of Outliers: เมืองขนาดยักษ์ (Megacity เช่น Tokyo) อาจส่งอิทธิพลต่อค่าสหสัมพันธ์แบบ Pearson ได้""")
````

4. เดือนใดของแต่ละภูมิภาคที่ควรหลีกเลี่ยงการเดินทางมากที่สุด
   ให้กำหนดเกณฑ์การตัดสินใจจากข้อมูลที่มีอยู่ และอธิบายเหตุผลของการเลือกเกณฑ์ดังกล่าว

````python
# คำตอบข้อ 18.4
# กำหนดเกณฑ์การตัดสินใจหลีกเลี่ยงการเดินทาง:
# 'ดัชนีความไม่เหมาะสมในการเดินทาง (Travel Discomfort Index)'
# เกณฑ์การตัดสินใจ:
# 1. ปริมาณฝนและจำนวนวันฝนตกหนัก (เป็นอุปสรรคสำคัญต่อการท่องเที่ยวและการคมนาคม)
# 2. อุณหภูมิสุดโต่ง (ร้อนจัด > 28 °C หรือ หนาวจัด < 15 °C)

monthly_travel_eval = (
    clean_data.groupby(["region", "month"])
    .agg(
        total_rain=("rainfall_mm", "sum"),
        heavy_rain_days=("rainfall_mm", lambda s: (s > 10).sum()),
        avg_temp=("temperature_c", "mean"),
    )
    .reset_index()
)

# คำนวณความเสี่ยงจากอุณหภูมิ (Temp penalty) และดัชนีรวม
temp_penalty = np.maximum(0, monthly_travel_eval["avg_temp"] - 28) + np.maximum(0, 15 - monthly_travel_eval["avg_temp"])
monthly_travel_eval["risk_score"] = (monthly_travel_eval["total_rain"] / 10.0) + (monthly_travel_eval["heavy_rain_days"] * 5.0) + (temp_penalty * 5.0)

# ชิ้นงาน: ตารางสรุปดัชนีความไม่เหมาะสม และเดือนที่ควรหลีกเลี่ยงที่สุด
pivot_risk = monthly_travel_eval.pivot_table(index="region", columns="month", values="risk_score").round(1)
worst_travel_months = monthly_travel_eval.loc[monthly_travel_eval.groupby("region")["risk_score"].idxmax()].reset_index(drop=True)

print("ตารางดัชนีความไม่เหมาะสมในการเดินทาง (คะแนนยิ่งสูงยิ่งควรหลีกเลี่ยง):")
print(pivot_risk)

print("\nสรุปเดือนที่ควรหลีกเลี่ยงการเดินทางมากที่สุดของแต่ละภูมิภาค:")
print(worst_travel_months[["region", "month", "total_rain", "heavy_rain_days", "avg_temp", "risk_score"]].round(2))

print("""\n[เหตุผลของการเลือกเกณฑ์และการอธิบาย]:
1. เหตุผลของการเลือกเกณฑ์:
   - ฝนตกหนักและปริมาณฝนสะสมสูง: เป็นปัจจัยลบอันดับหนึ่งของการเดินทางท่องเที่ยว ทำให้เกิดน้ำท่วม การจราจรติดขัด และกิจกรรมกลางแจ้งต้องยกเลิก
   - อุณหภูมิสุดโต่ง: อากาศที่ร้อนเกินไป (>28°C) หรือหนาวเกินไป (<15°C) ทำให้การเดินทางเหนื่อยล้าหรือไม่สบายตัว
2. ข้อสรุปรายภูมิภาค:
   - Asia: ควรหลีกเลี่ยง 'เดือน 2' มากที่สุด เนื่องจากมีฝนตกหนักสะสมสูงสุด (38 วันฝนตกหนัก)
   - Europe: ควรหลีกเลี่ยง 'เดือน 1' เนื่องจากเป็นช่วงที่อากาศหนาวเย็นที่สุดในไตรมาสแรก (เฉลี่ย 10.4 °C)
   - Africa: ควรหลีกเลี่ยง 'เดือน 1' เนื่องจากมีปริมาณฝนรวมสูงที่สุด (506.8 มม.)
   - North America: ควรหลีกเลี่ยง 'เดือน 1' เนื่องจากสภาพอากาศหนาวเย็นที่สุด
   - South America: ควรหลีกเลี่ยง 'เดือน 1' เนื่องจากมีฝนสะสมและวันฝนตกชุกที่สุดในไตรมาส""")
````
