# Part 1: Social Listening Warm-up
*งานเดี่ยว — ทุกคนต้องทำและส่งของตนเอง ไม่ต้องนำเสนอ (10 คะแนน)*

ใช้ข้อมูลโพสต์สาธารณะจาก **Bluesky**  
https://huggingface.co/datasets/alpindale/two-million-bluesky-posts

ใช้เพียง **1–3 ไฟล์** ก็พอ ไม่จำเป็นต้องโหลดทั้ง dataset

ฟิลด์ที่มี เช่น `text`, `created_at`, `author`, `uri`, `has_images`, `reply_to`, `predicted_language`

## ตอบ 4 คำถามหลัก

1. **คนโพสต์เมื่อไร:** จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
2. **ใช้ภาษาอะไร:** รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
3. **คนพูดถึงอะไร:** หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
4. **ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้:** สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด

> Part 1 เป็น warm-up ไม่ต้องทำ influencer analysis หรือ reply-network เว้นแต่สนใจทำเพิ่มเอง


```python
# ----------------- Your code here -----------------
import json
import os
import urllib.request
import urllib.error
import datetime
from collections import Counter
import re
import matplotlib.pyplot as plt

# Using langdetect for language detection
try:
    from langdetect import detect, DetectorFactory
    DetectorFactory.seed = 0
    HAS_LANGDETECT = True
except ImportError:
    HAS_LANGDETECT = False
    print("langdetect not installed, using regex heuristic for language detection")

DATA_DIR = "./data/"
os.makedirs(DATA_DIR, exist_ok=True)

url = "https://huggingface.co/datasets/alpindale/two-million-bluesky-posts/resolve/main/posts_20241127_075347.jsonl"
filename = "posts_20241127_075347.jsonl"
filepath = os.path.join(DATA_DIR, filename)

if not os.path.exists(filepath):
    print(f"Downloading {filename}...")
    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
    try:
        with urllib.request.urlopen(req) as response, open(filepath, 'wb') as out_file:
            out_file.write(response.read())
        print("Download complete.")
    except Exception as e:
        print(f"Failed to download: {e}")

def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)

posts = list(iter_jsonl(filepath))
print(f"Total posts loaded: {len(posts)}\n")

# 1. คนโพสต์เมื่อไร
timestamps = []
for p in posts:
    if 'created_at' in p:
        try:
            dt = datetime.datetime.fromisoformat(p['created_at'].replace('Z', '+00:00'))
            timestamps.append(dt)
        except Exception:
            pass

if timestamps:
    min_time = min(timestamps)
    max_time = max(timestamps)
    hours = [dt.hour for dt in timestamps]
    hour_counts = Counter(hours)
    peak_hour = hour_counts.most_common(1)[0][0]
    
    print("--- 1. คนโพสต์เมื่อไร ---")
    print(f"จำนวนโพสต์ทั้งหมด: {len(timestamps):,} โพสต์")
    print(f"ช่วงเวลาที่ครอบคลุม: {min_time.strftime('%Y-%m-%d %H:%M:%S')} ถึง {max_time.strftime('%Y-%m-%d %H:%M:%S')} (UTC)")
    print(f"ชั่วโมงที่มีการโพสต์มากที่สุด (UTC): {peak_hour:02d}:00 - {peak_hour:02d}:59 น. (จำนวน {hour_counts[peak_hour]:,} โพสต์)\n")

# 2. ใช้ภาษาอะไร
print("--- 2. ใช้ภาษาอะไร ---")
lang_counter = Counter()
for p in posts:
    text = p.get('text', '').strip()
    if not text:
        continue
    if 'predicted_language' in p:
        lang_counter[p['predicted_language']] += 1
    else:
        if HAS_LANGDETECT:
            try:
                lang = detect(text)
                lang_counter[lang] += 1
            except:
                lang_counter['unknown'] += 1
        else:
            if re.search(r'[\u0E00-\u0E7F]', text):
                lang_counter['th'] += 1
            elif re.search(r'[a-zA-Z]', text):
                lang_counter['en'] += 1
            elif re.search(r'[\u3040-\u309F\u30A0-\u30FF\u4E00-\u9FAF]', text):
                lang_counter['ja/zh'] += 1
            else:
                lang_counter['other'] += 1

total_langs = sum(lang_counter.values())
for lang, count in lang_counter.most_common(5):
    print(f"ภาษา {lang}: {count:,} โพสต์ ({count/total_langs*100:.1f}%)")
print("")

# 3. คนพูดถึงอะไร (Hashtag หรือ คำเด่น)
print("--- 3. คนพูดถึงอะไร ---")
hashtags = Counter()
for p in posts:
    text = p.get('text', '')
    tags = re.findall(r'#\w+', text)
    hashtags.update([t.lower() for t in tags])

print("Hashtags ที่พบบ่อยสุด:")
for tag, count in hashtags.most_common(5):
    print(f"{tag}: {count:,} ครั้ง")

print("\nตัวอย่างข้อความที่ใช้ Hashtag ยอดฮิต:")
top_tag = hashtags.most_common(1)[0][0] if hashtags else ""
found = 0
if top_tag:
    for p in posts:
        text = p.get('text', '')
        if top_tag in text.lower():
            print(f"- {text.replace(chr(10), ' ')}")
            found += 1
            if found >= 2: break
```

```python
# ----------------- Your code here -----------------
# 4. ข้อมูลบอกอะไรได้/ไม่ได้ (Visualization)
if timestamps:
    plt.figure(figsize=(10, 5))
    hours_sorted = sorted(hour_counts.items())
    plt.bar([x[0] for x in hours_sorted], [x[1] for x in hours_sorted], color='skyblue', edgecolor='black')
    plt.title('Bluesky Post Volume by Hour (UTC)', fontsize=14)
    plt.xlabel('Hour of day (UTC)', fontsize=12)
    plt.ylabel('Number of posts', fontsize=12)
    plt.xticks(range(24))
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Highlight peak hour
    plt.bar(peak_hour, hour_counts[peak_hour], color='salmon', edgecolor='black')
    
    plt.tight_layout()
    plt.show()
```

**<สรุปสั้น ๆ ถึงลูกค้า 5 บรรทัด>**

* **เวลาโพสต์:** ผู้ใช้ Bluesky โพสต์กระจุกตัวอย่างมากในช่วงเวลา 07:00-08:00 (UTC) (ประมาณช่วงบ่าย/เย็นในเอเชียและเช้าในอเมริกา) แสดงว่าเป็นช่วง Prime Time 
* **ภาษาหลัก:** ผู้ใช้ส่วนใหญ่ใช้ภาษาอังกฤษ (ประมาณ 61%) และภาษาญี่ปุ่น (ประมาณ 20%) ชี้ให้เห็นว่าแพลตฟอร์มนี้เป็นที่นิยมอย่างมากในกลุ่มผู้ใช้ญี่ปุ่นรองจากสหรัฐอเมริกา
* **หัวข้อยอดฮิต:** แฮชแท็กมักเกี่ยวข้องกับศิลปะ วัฒนธรรมป๊อป และเกม (เช่น #b3d, #precure, #vtuber) ซึ่งเป็นกลุ่มคอมมูนิตี้ที่เหนียวแน่นบน Bluesky
* **สิ่งที่มีประโยชน์จากข้อมูล:** ข้อมูลช่วยให้เราเข้าใจช่วงเวลาที่ดีที่สุดในการโพสต์ และกลุ่มเป้าหมายหลักบนแพลตฟอร์มว่าเป็นใครและสนใจเรื่องอะไร (ผ่าน hashtag)
* **ข้อจำกัดของข้อมูล:** ข้อมูลนี้เป็นพียงเสี้ยวหนึ่งของวัน (แค่หลักพันโพสต์ในช่วงแคบๆ) ยังไม่สามารถบอกเทรนด์รายสัปดาห์ การมีส่วนร่วมของโพสต์ หรือความรู้สึกผู้ใช้ (Sentiment) ได้อย่างลึกซึ้ง
